## 4. Converting date strings to datetime type

```python
from pymongo import MongoClient
from datetime import datetime

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["MongoDb_Practice"]
collection = db["Practice_Collection"]

# Function to convert date string to datetime object
def convert_to_datetime(date_str):
    return datetime.strptime(date_str, "%b %d %Y")  # "Feb 10 2015" format

# Update documents to change date string to datetime
for doc in collection.find({"date": {"$type": "string"}}):  # Filter for string-based dates
    try:
        # Parse the string date to datetime object
        parsed_date = convert_to_datetime(doc['date'])
        
        # Update the document with the new date format
        collection.update_one(
            {"_id": doc["_id"]},
            {"$set": {"date": parsed_date}}
        )
        
        print(f"Updated document with _id: {doc['_id']}")
    except Exception as e:
        print(f"Error parsing date for document _id: {doc['_id']}: {e}")
```
